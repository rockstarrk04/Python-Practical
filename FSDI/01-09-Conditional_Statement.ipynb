24. Write a Python program to identify whether a triangle is equilateral, isosceles, or scalene based on its three sides.

```python
side1 = int(input("Enter Side 1 : "))
side2 = int(input("Enter Side 2 : "))
side3 = int(input("Enter Side 3 : "))

if side1 + side2 + side3 == 180:
    if side1 == side2 == side3 :
        print("Equilateral Triangle")
    elif ((side1 == side2) and (side3 != side1)) or (side2 == side3) and (side3 != side1) :
        print("Isoceles Triangle")
    elif side1 != side2 != side3:
        print("Scalene Triangle")
else:
    print("It is not a Triangle")
```

30. Write a Python program for a simple calculator using conditional statements. Accept two numbers and an operator (+, -, *, /) and perform the corresponding operation.

```python
print("1.Addition")
print("2.Subtraction")
print("3.Multiplication")
print("4.Division")

num1 = int(input("Enter Number 1 : "))
num2 = int(input("Enter Number 2 : "))

choice = int(input("Enter your choice : "))

match (choice):
    case 1 :
        print("Addition")
        print(num1 + num2)
    case 2 :
        print("Subraction")
        print(num1 - num2)
    case 3 : 
        print("Multiplication")
        print(num1 * num2)
    case 4:
        print("Division")
        print(num1 / num2)
    case _:
        print("Invalid option")
```
