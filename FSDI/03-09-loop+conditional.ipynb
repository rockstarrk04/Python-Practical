2. Write a Python program to check whether a given number is an Armstrong number using only a while loop.

```python
num = int(input("Enter number to check Armstrong or Not : "))
temp = num
l = len(str(num))
sum = 0

while num > 0:
    last_digit = num % 10
    sum = sum + last_digit ** l
    num = num // 10

if temp == sum:
    print(f"{temp} is Armstrong")
else:
    print(f"{temp} is not Armstrong")
```

3. Write a Python program to check whether a given number is a perfect number using only a while loop.

```python
num = int(input("Enter number to check Perfect Number : "))
sum = 0

i=1
limit = num // 2
while i <= limit :
    if num % i == 0:
        sum = sum + i
    i = i+1
if sum == num:
    print(f"{num} is perfect Number")
else:
    print(f"{num} is not a perfect number")
```

4. Write a Python program to check whether a given number is a strong number using only a while loop.

```python
# Sum of each digit's factorial == orginal number
# 1. get last digit
# 2. find the factorial value of that number and add to sum
# 3. update number
# 4. repeat process

num = int(input("Enter number to check Strong Number or Not : "))
temp = num
sum = 0


while num > 0:
    last_digit = num % 10
    fact = 1
    i = 1
    while i <= last_digit:
        fact = fact * i
        i = i+1
    sum = sum + fact
    num = num // 10

if temp == sum:
    print(f"{temp} is Strong number")
else:
    print(f"{temp} is not a Strong Number")
```

9. Write a Python program to find the factorial of a given number using only a while loop.

```python
num = int(input("Enter a number :"))
fact = 1
if num == 1:
    print(f"Fact of {num} is {fact}")
else:
    i = 1
    while i <= num:
        fact = fact * i 
        i = i + 1

    print(f"Fact of {num} is {fact}")
```

10. Write a Python program to print all prime numbers between 1 and N using only while loops.

```python
num = int(input("Enter a number : "))
i = 1

while i <= num:
    count = 0
    j = 1
    while j <= i // 2:
        if i % j == 0:
            count = count + 1
        j = j + 1
    if count == 1:
        print(i, end = " ")
    i = i + 1
```

14. Write a Python program to count the number of uppercase letters, lowercase letters, digits, and special characters in a string using only a while loop.

```python
str = input("Enter a string : ")
U_count = 0
L_count = 0
Digit_count = 0
Special_count = 0

i = 0
while i < len(str):
    if ord(str[i]) >= ord('A') and ord(str[i]) <= ord('Z'):
        U_count += 1
    elif ord(str[i]) >= ord('a') and ord(str[i]) <= ord('z'):
        L_count += 1
    elif ord(str[i]) >= ord('0') and ord(str[i]) <= ord('9'):
        Digit_count += 1
    else:
        Special_count += 1
    i += 1

print(str)
print(f"Uppercase count : {U_count}")
print(f"Lowercase count : {L_count}")
print(f"Digit count : {Digit_count}")
print(f"Special Characters count : {Special_count}")
```

15. Write a Python program to count the frequency of each character in a given string using only while loops.

```python
str = input("Enter a sentence : ")

frequency = {}

i = 0
while i < len(str):
    if str[i].lower() != " ":
        if str[i].lower() in frequency:
            frequency[str[i].lower()] += 1
        else:
            frequency[str[i].lower()] = 1

    i += 1

print(frequency.items())
```

19. Write a Python program to find the longest word in a given string using only while loops.

```python
str = input("Enter a sentence to find longest word : ")
array = []

l_word = ""
count = 0

array = str.split()
i = 0
while i < len(array):
    length = len(array[i])
    if length > len(l_word) :
        l_word = array[i]
        count = length

    i += 1

print(str)
print(f"Longest Word : {l_word}")
print(f"Length of Longest Word : {count}")
```

20. Write a Python program to remove duplicate characters from a string using only while loops.

26. Write a Python program to reverse a list using only a while loop.

```python
numbers = [20, 23, 40, 34, 32, 78, 92]
left = 0
right = len(numbers) - 1
print(numbers)
while left < right:
    temp = numbers[left]
    numbers[left] = numbers[right]
    numbers[right] = temp

    left += 1
    right -= 1

print(numbers)
```

30. Write a Python program to separate positive, negative, and zero elements from a list using only while loops.

```python
numbers = [-20, 0 , 23, -40, 34, 32, -78, 92 , 0]
positive = []
negative = []
zero = []

i = 0
while i < len(numbers):
    if numbers[i] > 0:
        positive.append(numbers[i])
    elif numbers[i] < 0:
        negative.append(numbers[i])
    elif numbers[i] == 0:
        zero.append(numbers[i])

    i += 1
print(positive)
print(negative)
print(zero)
```
