**WAP to find the position of 3rd occurrance of digit 20 in given tuple collection**

```python
t1 = (10,20,30,30,20,30,20,98)
value = 20
count = 0
for index , digit in enumerate(t1):
    if digit == value:
        count = count + 1
        if count == 3:
            print(index)
            break
```
