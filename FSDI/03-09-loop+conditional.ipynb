2. Write a Python program to check whether a given number is an Armstrong number using only a while loop.

```python
num = int(input("Enter number to check Armstrong or Not : "))
temp = num
l = len(str(num))
sum = 0

while num > 0:
    last_digit = num % 10
    sum = sum + last_digit ** l
    num = num // 10

if temp == sum:
    print(f"{temp} is Armstrong")
else:
    print(f"{temp} is not Armstrong")
```

3. Write a Python program to check whether a given number is a perfect number using only a while loop.

```python
num = int(input("Enter number to check Perfect Number : "))
sum = 0

i=1
limit = num // 2
while i <= limit :
    if num % i == 0:
        sum = sum + i
    i = i+1
if sum == num:
    print(f"{num} is perfect Number")
else:
    print(f"{num} is not a perfect number")
```

4. Write a Python program to check whether a given number is a strong number using only a while loop.

```python
# Sum of each digit's factorial == orginal number
# 1. get last digit
# 2. find the factorial value of that number and add to sum
# 3. update number
# 4. repeat process

num = int(input("Enter number to check Strong Number or Not : "))
temp = num
sum = 0


while num > 0:
    last_digit = num % 10
    fact = 1
    i = 1
    while i <= last_digit:
        fact = fact * i
        i = i+1
    sum = sum + fact
    num = num // 10

if temp == sum:
    print(f"{temp} is Strong number")
else:
    print(f"{temp} is not a Strong Number")
```

6. Write a Python program to find the sum of digits of a given number using only a while loop.

```python
num = int(input("Enter a number :"))
sum = 0
while num > 0:
    last_digit = num % 10
    sum = sum + last_digit

    num  = num // 10
print(f"Sum of digits : {sum}") 
```

7. Write a Python program to reverse a given number using only a while loop.

```python
num = int(input("Enter a number :"))
temp = num 
rev = 0

while num > 0:
    last_digit = num % 10
    rev = rev * 10 + last_digit
    num = num // 10
print(f"reversed Number of {temp} is {rev}")
```

8. Write a Python program to count the number of digits in a given number using only a while loop.

```python
num = int(input("Enter a number :"))
n = num
count = 0

while num > 0:
    count = count + 1
    num = num // 10
print(f"Count of {n} is {count}")
```

9. Write a Python program to find the factorial of a given number using only a while loop.

```python
num = int(input("Enter a number :"))
fact = 1
if num == 1:
    print(f"Fact of {num} is {fact}")
else:
    i = 1
    while i <= num:
        fact = fact * i 
        i = i + 1

    print(f"Fact of {num} is {fact}")
```
