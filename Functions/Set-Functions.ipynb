***add method***

```python
set = {10,20,30,40,50}
set.add(90)
print(set)
```

***remove method***

```python
set = {10,20,30,40,50}
set.remove(40)
print(set)
```

***update method***

```python
set1 = {10,20,30,40,50}
set2 = {20,40,50,90,100}
set1.update(set2)    # set with set
print(set1)
```

```python
set1 = {10,20,30,40,50}
list = [20,40,50,90,100]
set1.update(list)   # set with list
print(set1)
```

```python
set1 = {10,20,30,40,50}
tuple = (20,40,50,90,100)
set1.update(tuple)  # set with tuple
print(set1)
```

```python
set1 = {10,20,30,40,50}
r = range(5)
set1.update(r)   # set with range
print(set1)
```

```python
set1 = {10,20,30,40,50}
string = 'python'
set1.update(string)     # set with string
print(set1)
```

***pop method***

```python
set1 = {10,20,30,40,50,59}
set1.pop()  
print(set1)    # output cannot be predictable
```

***clear method***

```python
set1 = {10,20,30,40,50,59}
set1.clear()  
print(set1)
```

***union method***

```python
set1 = {10,20,30,40,50}
set2 = {20,40,50,90,100}
r = set1.union(set2)    
print(r)
```

***intersection method***

```python
set1 = {10,20,30,40,50}
set2 = {20,40,50,90,100}
r = set1.intersection(set2)    
print(r)
```

***symmetric_difference method***

```python
set1 = {10,20,30,40,50}
set2 = {20,40,50,90,100}
r = set1.symmetric_difference(set2)    
print(r)
```

***difference method***

```python
set1 = {10,20,30,40,50}
set2 = {20,40,50,90,100}
r1 = set1.difference(set2)    
r2 = set2.difference(set1)
print(r1)
print(r2)
```
