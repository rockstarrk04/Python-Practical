```python
# with argument without return
def method_name(name):
    print(f"name is {name}")


method_name('santhanam')
```

```python
# with argument with return
def method_name(a,b):
    return a+b
    
sum = method_name(5,10)
print(sum)
```

```python
# without argument with return
def method_name():
    a = 5
    b = 10
    res = a + b
    return res

result = method_name()
print(result)
```

```python
# without argument without return
def method_name():
    a = 5
    b = 10
    res = a + b
    print(res)

method_name()
```
