***WAP to print even , odd numbers separately from a given list collection using append method***

```python
print('print even , odd numbers separately from a given list collection using append method')
li = [10,23,25,20,40,98,43,44]
even_li = []
odd_li = []
for i in li:
    if i % 2 == 0:
        even_li.append(i)
    else:
        odd_li.append(i)
print(f'Even numbers in list : {even_li}')
print(f'Odd numbers in list : {odd_li}')
print()
```

***WAP to print sum of  even ,sum of  odd numbers separately from a given list collection.***


```python
print('print sum of  even ,sum of  odd numbers')
li = [10,23,25,20,40,98,43,44]
even_li = []
odd_li = []
even_sum = 0
odd_sum = 0

for i in li:
    if i % 2 == 0:
        even_li.append(i)
    else:
        odd_li.append(i)
for i in even_li:
    even_sum += i
for i in odd_li:
    odd_sum += i

print(f'Sum of even numbers : {even_sum}')
print(f'Sum of odd numbers : {odd_sum}')
```

```python
li = [40,23,40,20,40,98,43,44]
first = li.index(40)
second = li.index(40,first + 1)
third = li.index(40, second + 1 )
print(third)
```

```python
# Using enumerate method

li = [40,23,40,20,40,98,43,44]
count = 0
for index , digit in enumerate(li):
    if digit == 40:
        count += 1
        if count == 3:
            print(index)
            break
```

***Reverse method***

```python
li = [40,23,40,20,40,98,43,44]
li.reverse()
print(li)
```

***sort method***

```python
# Based on value and in Ascending order(default)

li = [40,23,40,20,40,98,43,44]
li.sort()     # or li.sort(reverse=False)     
print(li)
```

```python
# Based on value and in Descending order

li = [40,23,40,20,40,98,43,44]
li.sort(reverse=True)
print(li)
```

```python
# Based on Unicode (ASCII) value of first Character of each item
# default in Ascending order

words = ['ram', 'bala' , 'Nithin' , 'Bala']
words.sort()
print(words)
```

```python
# Based on length of each item
# default in Ascending order

words = ['ram', 'bala' , 'Nithin' , 'Bala']
words.sort(key=len)
print(words)
```
