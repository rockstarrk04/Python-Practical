**Membership Operator**

```python
s1 = 'python'
print('p' in s1)
print('P' in s1)
```

```python
s1 = 'Hello students welcome to python class'
print('HELLO' in s1)
print('welcome' in s1)
print('Java' in s1)
print('TO' in s1)
```

```python
list = [10,20,30]
print(10 in list)
print(11 in list)
print(30 in list)
print(int('20') in list)
print(int('11') in list)
```

```python
tuple = (10,20,30,40)
print(10 in tuple)
print(11 in tuple)
print(30 in tuple)
print(int('20') in tuple)
print(int('11') in tuple)
```

```python
set = {10,20,30,40}
print(10 in set)
print(11 in set)
print(30 in set)
print(int('20') in set)
print(int('11') in set)
```

```python
set = {'RAM' , 'NITHIN', 20 , 10 }
print('ram' in set)
print('RAM' in set)
print(20 in set)
print(11 in set)
```

```python
dict = {
    'ename' : 'smith',
    'empno' : 7839
}
print('ename' in dict)
print('Ename' in dict)
print('sal' in dict)
print(7839 in dict.values())
print(123 in dict.values())
```

```python
dict1 = {
    'ename' : 'smith',
    'sal' : 800
}
print(800 in dict1.keys())
print(800 in dict1.values())
print('smith' in dict1.values())
print('smith' not in dict1.values())
```

**Nested Dictionary**

```python
nested_dict = {
    'employee1':{
        'ename' : 'John Smith',
        'sal' : 80000
    },
    'employee2':{
        'ename' : 'Jane Doe',
        'sal' : 90000
    },
    'employee3':{
        'ename' : 'Alice Johnson',
        'sal' : 75000
    }
}

print('employee1' in nested_dict) # True
print('employee1' not in nested_dict)   # False
print('employee1' in nested_dict.keys()) # True
print('employee4' in nested_dict) # False
print('ename' in nested_dict['employee1']) # True 
print('sal' in nested_dict['employee1']) # True
print('empno' in nested_dict['employee1'])  # False
print(75000 in nested_dict['employee3'].values()) #True
print(('ename','Jane Doe') in nested_dict['employee2'].items()) # True
```
