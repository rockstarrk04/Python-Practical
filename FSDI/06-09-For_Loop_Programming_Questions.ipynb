8. Write a Python program to find the factorial of a given number using a for loop.

```python
num = int(input("Enter a number to find it's Factorial : "))
fact = 1
if num == 0 or num == 1:
    print(f"Fact of {num} is 1")
else:
    for i in range(1,num+1):
        fact = fact * i
    print(f"Fact of {num} is {fact}")
```

11. Write a Python program to reverse a given number using a for loop.

```python
num = int(input("Enter a number : "))

temp = abs(num)
rev = 0

for i in range(len(str(temp))):
    digit = temp % 10
    rev = rev * 10 + digit
    temp = temp // 10

if num < 0:
    rev = -rev

print(f"Number : {num}")
print(f"Reverse : {rev}")
```

12. Write a Python program to check whether a given number is a palindrome using a for loop.

```python
num = int(input("Enter a number : "))
rev = ""
for i in str(num):
    rev = i + rev

if int(rev) == num:
    print('palindrome')
else:
    print('Not a palindrome')
```

13. Write a Python program to check whether a given number is an Armstrong number using a for loop.

```python
num = int(input("Enter a number : "))
sum = 0
length = len(str(num))

for i in str(num):
    sum = sum + int(i) ** length

if sum == num:
    print('Armstrong')
else:
    print('Not a armstrong')
```

23. Write a Python program to count the number of uppercase and lowercase characters in a string using a for loop.

```python
string = input('Enter a string : ')
uppercase_count = 0
lowercase_count = 0

for i in string:
    if ord(i) >= ord('A') and ord(i) <= ord('Z'):
        uppercase_count += 1
    elif ord(i) >= ord('a') and ord(i) <= ord('z'):
        lowercase_count += 1

print(string)
print(f"Uppercase count : {uppercase_count}")
print(f"Lowercase count : {lowercase_count}")
```
