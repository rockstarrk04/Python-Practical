```python
#  general copy method
num1 = [1,2,3,4]
num2 = num1
num2[0] = 99 
print(num1)
print(num2)
```

```python
#  shallow copy method
import copy
# flat list
num1 = [1,2,3,4]
num2 = copy.copy(num1) 
num2[0] = 99
print(f' original {num1}')
print(f' copy {num2}')
```

```python
#  shallow copy method
import copy
# nested list
num1 = [1,[10,20,30],2,3,4]
num2 = copy.copy(num1) 
num1[1][0] = 99
print(f' original {num1}')
print(f' copy {num2}')
```

```python
# Deep copy method
import copy
# flat list
num1 = [1,2,3,4]
num2 = copy.deepcopy(num1) 
num2[0] = 99
print(f' original {num1}')
print(f' copy {num2}')
```

```python
# Deep copy method
import copy
# Nested list
num1 = [1,[10,20,30],2,3,4]
num2 = copy.deepcopy(num1) 
num1[1][0] = 99
print(f' original {num1}')
print(f' copy {num2}')
```
