**ZeroDivisionError**

```python
try:
    a = 100
    b = 0
    result = a/b
    print(result)
except:
    print('Enter valid number to perform division')
```

**IndexError**

```python
try :
    num = [1,2,3,4,5]
    print(num[10])
except : 
    print('Enter a valid index position')
```

**TypeError**

```python
try:
    str = 'ram'
    num = 1234
    result = str + num 
    print(result)
except:
    print('TypeError: can only concatenate str (not "int") to str')
```
