```python
num = int(input('Enter number of times to print 1 to : '))
sum = 0
i = 1
while i <= num:
    if i % 2 == 0:
        sum = sum + i
    i = i + 1
print(f'Sum of 1 to n is : {sum}')
```

```python
n = 11
count = 0

while n > 0:
    if n % 10 == 0:
        count = count + 1
    n = n // 10
if count >= 1:
    print('duck')
else:
    print('no duck')
```

```python
## Automorphic Number

n = 25
count_n = n
count = 0 # 2
while count_n > 0:
    count = count + 1
    count_n = count_n // 10 

sq = n * n # 625
val =sq % (10 ** count) # 100
if val == n:
    print('Automorphic Number')
else:
    print('not Automorphic Number')
```

```python
s = 'AEIOUaeiou'
uppercase = 0
lowercase = 0
i = 0

while i < len(s):
    if s[i] >= 'A' and s[i] <= 'Z':
        uppercase = uppercase + 1
    elif s[i] >= 'a' and s[i] <= 'z':
        lowercase = lowercase + 1
    i = i + 1
    
print(f'Uppercase count: {uppercase}')
print(f'Lowercase count: {lowercase}') 
```

```python
s = '123AEIOU@!#qwrt'
vowel_count = 0
consonant_count = 0
digit_count = 0

i = 0

while i < len(s):
    if s[i] in 'AEIOUaeiou':
        vowel_count = vowel_count + 1
    elif s[i] >= 'A' and s[i] <= 'z':
        if s[i] not in 'AEIOUaeiou':
            consonant_count = consonant_count + 1
    elif s[i] >= '0' and s[i] <= '9':
        digit_count = digit_count + 1

    i = i + 1
print(f'Vowel count : {vowel_count}')
print(f'Consonant count : {consonant_count}')
print(f'Digit count : {digit_count}')
```

```python
s = 'Welcome to python class'
input = s.split()
i = 0
longest_word = ''
while i < len(input):
    if len(input[i]) > len(longest_word):
        longest_word = input[i]
    i = i + 1
print(longest_word)
```

```python
s = input("Enter a string: ")

result = ""
i = 0

while i < len(s):
    if s[i] >= 'a' and s[i] <= 'z':
        result = result + chr(ord(s[i]) - 32)
    else:
        result = result + s[i]

    i = i + 1

print(result)
```

```python
print("Welcome to Book My Show!")

while True: 
    theater_name = input("\nEnter the name of the theater (PVR, INOX) or type 'exit' to quit: ")

    if theater_name.lower() == "exit":
        print("Thank you for using Book My Show!")
        break

    if theater_name == "PVR" or theater_name == "INOX":

        if theater_name == "PVR":
            print("\nMovies available at PVR:")
            movie1 = "Avatar: The Way of Water - RS 850"
            movie2 = "The Batman - RS 800"
            ticket_price1 = 850
            ticket_price2 = 800
        else:
            print("\nMovies available at INOX:")
            movie1 = "Mission: Impossible - Fallout - RS 900"
            movie2 = "Jurassic World: Dominion - RS 950"
            ticket_price1 = 900
            ticket_price2 = 950

        print(f"1. {movie1}")
        print(f"2. {movie2}")

        movie_choice = input("\nEnter the number of your chosen movie (1 or 2): ")

        if movie_choice == "1" or movie_choice == "2":

            if movie_choice == "1":
                selected_movie = movie1
                ticket_price = ticket_price1
            else:
                selected_movie = movie2
                ticket_price = ticket_price2

            print(f"\nYou have selected: {selected_movie}")
            print(f"Ticket Price: RS {ticket_price}")

            confirmation = input("\nDo you want to confirm the booking? (yes/no): ")

            if confirmation == "yes":
                print(f"\nYour ticket for '{selected_movie}' has been booked successfully!")
            else:
                print("\nTicket booking has been canceled.")

        else:
            print("Invalid movie choice. Please try again.")

    else:
        print("Sorry, the theater name is not available.")
```
